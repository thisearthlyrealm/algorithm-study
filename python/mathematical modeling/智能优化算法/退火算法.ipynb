```python
import math
import random
```

首先导入必备的算法库

```python
def euclidean_distance(city1,city2):
    return math.hypot(city1[0]-city2[0],city1[1]-city2[1])
```

工具函数：（1）计算两城市之间的欧氏距离

```python
def compute_total_distance(tour,dist_matrix):
    total=0.0
    n=len(tour)
    for i in range(n):
        total+=dist_matrix[tour[i]][tour[(i+1)%n]]
    return total
```

工具函数：（2）计算一条路径（哈密顿回路）的总长度

```python
def two_opt_swap(tour,i,j):
    new_tour=tour[:i]+tour[i:j+1][::-1]+tour[j+1:]
    return new_tour
```

邻域生成：（1）2—opt交换：反转tour中i-j之间的片段，返回一个新tour。

```python
def generate_neighbor(tour):
    n=len(tour)
    if random.random()<0.8:
        i,j=random.sample(range(n),2)
        if i>j:
            i,j=j,i
        return two_opt_swap(tour,i,j)
    else:
        i,j=random.sample(range(n),2)
        new_tour=tour[:]
        new_tour[i],new_tour[j]=new_tour[j],new_tour[i]
        return new_tour
```

邻域生成：（2）随机生成一个邻域解：以80%概率使用2—opt,20%概率使用交换两个城市。

```python
def simulated_annealing_tsp(dist_matrix,initial_tour,T0=1000.0,T_min=1e-8,alpha=0.999,L=200):
    len(initial_tour)
    current_tour=initial_tour[:]
    current_length=compute_total_distance(current_tour,dist_matrix)
    best_tour=current_tour[:]
    best_length=current_length
    T=T0
    iteration=0
    while T>T_min:
        for _ in range(L):
            new_tour=generate_neighbor(current_tour)
            new_length=compute_total_distance(new_tour,dist_matrix)
            delta=new_length-current_length
            if delta<0 or random.random()<math.exp(-delta/T):
                current_tour=new_tour[:]
                current_length=new_length
                if current_length<best_length:
                    best_length=current_length
                    best_tour=current_tour[:]
        T*=alpha
        iteration+=1
        if iteration%800==0:
            print(f"总退火代数:{iteration}")
    print(f"总退火代数:{iteration}")
    return best_tour,best_length
```

模拟退火核心：使用模拟退火求解TSP：
（1）dist_matrix：距离矩阵
（2）initial_tour：初始路径
（3）T0：初始温度
（4）T_min：终止温度
（5）alpha：降温系数
（6）L：每个温度下内循环次数
（7）return：（最优路径，最优长度）

```python
def main():
    num_cities=20
    random.seed(100)
    cities=[(random.uniform(0,1000),random.uniform(0,1000)) for _ in range(num_cities)]
    dist_matrix=[[0.0]*num_cities for _ in range(num_cities)]
    for i in range(num_cities):
        for j in range(num_cities):
            dist_matrix[i][j]=euclidean_distance(cities[i],cities[j])
    initial_tour=list(range(num_cities))
    random.shuffle(initial_tour)
    T0=5000.0
    T_min=1e-10
    alpha=0.995
    L=300
    best_tour,best_length=simulated_annealing_tsp(dist_matrix,initial_tour,T0=T0,T_min=T_min,alpha=alpha,L=L)
    print(f"城市数目:{num_cities}")
    print(f"最优路径:{best_tour}")
    print(f"最优距离:{best_length:.4f}")
    path_coords=[cities[i] for i in best_tour]+[cities[best_tour[0]]]
if __name__=="__main__":
    main()
```

主函数：
（1）生成城市。
（2）计算距离矩阵。
（3）初始解。
（4）参数设置和运行模拟退火。
（5）输出结果。
